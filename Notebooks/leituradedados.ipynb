```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DATA_DIR = Path("..") / "Data"

candidatos = pd.read_csv(
    DATA_DIR / "presidente_sc_ficticio_candidatos.csv",
    sep=";",
    encoding="utf-8-sig",
    decimal=",",
)
resumo = pd.read_csv(
    DATA_DIR / "presidente_sc_ficticio_resumo.csv",
    sep=";",
    encoding="utf-8-sig",
)

print(f"{len(resumo)} municípios e {len(candidatos)} registros carregados.")
display(candidatos.head())

# Total de votos por candidato.
votos_candidato = (
    candidatos.groupby("candidato")["votos"]
    .sum()
    .sort_values(ascending=True)
)

ax = votos_candidato.plot.barh(figsize=(10, 5), color="#2f6f9f")
ax.set_title("Total de votos por candidato")
ax.set_xlabel("Votos")
ax.set_ylabel("")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()

# Comparecimento e abstenção nos 10 maiores eleitorados.
maiores = resumo.nlargest(10, "eleitorado").sort_values("eleitorado")
taxas = maiores.assign(
    comparecimento_pct=maiores["comparecimento"] / maiores["eleitorado"] * 100,
    abstencao_pct=maiores["abstencao"] / maiores["eleitorado"] * 100,
).set_index("municipio")

ax = taxas[["comparecimento_pct", "abstencao_pct"]].plot.barh(
    figsize=(10, 6),
    color=["#3a9d5d", "#d95f59"],
)
ax.set_title("Comparecimento e abstenção nos 10 maiores eleitorados")
ax.set_xlabel("Percentual do eleitorado (%)")
ax.set_ylabel("")
ax.legend(["Comparecimento", "Abstenção"])
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()

# Municípios com maior percentual de abstenção.
tabela = resumo.assign(
    comparecimento_pct=resumo["comparecimento"] / resumo["eleitorado"] * 100,
    abstencao_pct=resumo["abstencao"] / resumo["eleitorado"] * 100,
)[["municipio", "eleitorado", "comparecimento_pct", "abstencao_pct", "mais_votado"]]

display(
    tabela.sort_values("abstencao_pct", ascending=False)
    .head(10)
    .style.format({"comparecimento_pct": "{:.1f}%", "abstencao_pct": "{:.1f}%"})
)
```
